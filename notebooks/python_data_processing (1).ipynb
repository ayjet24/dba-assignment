**Import Python Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

**Load CSV files from GitHub**

```python
base = "https://raw.githubusercontent.com/ayjet24/dba-assignment/main/data/raw_csv_files/"
```

**Read CSV files / convert them into pandas DataFrames**

```python
app_events = pd.read_csv (base + "app_events.csv")
complaints = pd.read_csv (base + "complaints.csv")
customers = pd.read_csv (base + "customers.csv")
data_dictionary = pd.read_csv (base + "data_dictionary.csv")
deliveries = pd.read_csv (base + "deliveries.csv")
drivers = pd.read_csv (base + "drivers.csv")
hubs = pd.read_csv (base + "hubs.csv")
incidents = pd.read_csv (base + "incidents.csv")
orders = pd.read_csv (base + "orders.csv")
vehicles = pd.read_csv (base + "vehicles.csv")
```

**Display the output**

```python
print("app_events:", app_events.shape)
print("complaints:", complaints.shape)
print("customers:", customers.shape)
print("data_dictionary:", data_dictionary.shape)
print("deliveries:", deliveries.shape)
print("drivers:", drivers.shape)
print("hubs:", hubs.shape)
print("incidents:", incidents.shape)
print("orders:", orders.shape)
print("vehicles:", vehicles.shape)
```

**Store datasets in a single dictionary**

```python
dataframes = {
    "app_events": app_events,
    "complaints": complaints,
    "customers": customers,
    "data_dictionary": data_dictionary,
    "deliveries": deliveries,
    "drivers": drivers,
    "hubs": hubs,
    "incidents": incidents,
    "orders": orders,
    "vehicles": vehicles
}
```

**Each file will be looked into further**

```python
for name, df in dataframes.items():
  print ("DATASET:", name)
  display(df.head())
  print(df.info())
  print("Values that are missing:")
  print(df.isnull().sum())
  print("Duplicates:", df.duplicated().sum())
```

**Column names will be cleaned**

```python
for name, df in dataframes.items():
  columns_clean = []

  for col in df.columns:

    col = col.strip() # spaces are removed before and after the text

    col = col.lower() # letters become lowercase

    col = col.replace (" ", "_") # spaces are replaced with underscores

    col = col.replace ("-", "_") # hyphens are replaced with underscores

    columns_clean.append(col) # add cleaned column name to the list

# cleaned column names are put back into the dataframe
df.columns = columns_clean

# save the dataframe in the dictionary
dataframes[name] = df
```

**Make text columns consistent**

```python
for name, df in dataframes.items(): # lopp through each dataframe
  for col in df.columns: # lopp through each column

    if df[col].dtype == "object": # see if column is text
      df[col] = df[col].astype(str) # string conversion
      df[col] = df[col].str.strip() # extra spaces are removed
      df[col] = df[col].str.lower() # make lowercase

dataframes[name] = df # save the updated dataframe
```

**Check the missing values**

```python
for name in dataframes: # loop through each dataset within the dictionary

  df = dataframes[name] # retrieve the DataFrame

  print()

  print ("Name of the dataset:", name) # print the name of the dataset

  values_missing = df.isnull().sum() # locate the missing values in each column

  print ("The values that are missing in each column:")
  print(values_missing)
```

**Cleaning the data**

```python
for name, df in dataframes.items(): # loop through each dataset

  for col in df.columns: # loop through each column

    if df[col].dtype == "object": # if the column has text data

      df[col] = df[col].replace("nan", "not known")
      df [col] = df[col].fillna("not known") # fill the missing text with "not known"

    else: # if the column includes numerical data
      df[col] = df[col].fillna(df[col].median())

  dataframes[name] = df # save cleaned dataset
```

**Locate columns that are important**

```python
for name, df in dataframes.items():

  print()

  print("Name of the Dataset:", name) # print the dataset name

  print ("Columns:", df.columns.tolist()) # print columns names as a list
```

**Display CLEANED Data**

```python
for name, df in dataframes.items():

  print("CLEANED DATASET", name) # print dataset name

  display(df.head()) # display first 5 rows from the dataset

  print("These are the missing values are cleaning:")
  print(df.isnull().sum()) # try to find missing values in each column

  print("Duplicates:")
  print(df.duplicated().sum()) # find duplicate rows
```

**Chart for analysis**

**Battery health of vehicle by maintenance status**

```python
# 1. the data should be grouped by maintenance status
# 2. the 'battery_health_pct' column should be looked at
# 3. the average battery health for each maintenance status is calculated

vehicles = dataframes ["vehicles"]

battery_health_maintenance = vehicles.groupby("maintenance_status")["battery_health_pct"].mean()

print(battery_health_maintenance) # print the result so the average can be viewed

battery_health_maintenance.plot(kind="bar") # bar chart is created

plt.title("The Battery Health by Maintenance Status - Average") # title for the chart

plt.ylabel ("Battery Health Average in %") # print the y-axis

plt.xlabel ("Status of maintenance") # print the x-axis

plt.show() # display the final chart
```

**Chart for analysis**

**Complaints by complaint type**

```python
# 1. View the "complaint_type" column
# 2. Count the number of times each complaint type is shown

complaints = dataframes ["complaints"]

numberof_complaint_type = complaints ["complaint_type"].value_counts()

print (numberof_complaint_type) # print the result so the number of complaints can be viewed

numberof_complaint_type.plot(kind="bar") # draw a chart

plt.title ("Complaints by Type of Complaint") # title for the chart

plt.ylabel ("Number of Complaints") # print the y-axis

plt.xlabel ("Type of Complaint") # print the x-axis

plt.show() # display the final chart
```

**Chart for analysis**

**Average delivery rating by delivery status**

```python
# 1. View the "delivery_status" column
# 2. View the "customer_rating_post_delivery" column
# 3. Group all the rows with similar delivery status
# 4. Identify the average ratings from the customer for each delivery status

deliveries = dataframes ["deliveries"]

rating_average_status = deliveries.groupby("delivery_status")["customer_rating_post_delivery"].mean()

print (rating_average_status) # print the result so the average can be viewed

rating_average_status.plot(kind="bar") # draw a chart

plt.title ("Average Delivery Rating by Delivery Status") # title for the chart

plt.ylabel ("Average Rating") # print the y-axis

plt.xlabel ("Delivery Status") # print the x-axis

plt.show() # display the final chart
```

**Chart for analysis**

**Delivery status count**

```python
# 1. View the "delivery_status" column
# 2. Count the number of times each delivery status shows up

deliveries = dataframes["deliveries"]

numberof_delivery_status = deliveries["delivery_status"].value_counts()

print (numberof_delivery_status) # result is printed so the number of deliveries can be seen

numberof_delivery_status.plot(kind="bar") # draw a chart

plt.title ("Delivery Status Count") # title for the chart

plt.ylabel ("Number of Deliveries") # print the y-axis

plt.xlabel ("Delivery Status") # print the x-axis

plt.show() # display
```

**Joined analysis**

```python
# retrieve the tables needed
orders = dataframes ["orders"]
deliveries = dataframes ["deliveries"]

# combine the orders and deliveries using order_id
deliveries_order = deliveries.merge(
    orders[["order_id", "service_type", "priority_level", "order_value"]],
    on="order_id",
    how="left"
)

# count the number of deliveries in each status for each type of service

status_service = deliveries_order.groupby(
    ["service_type", "delivery_status"]
).size().unstack(fill_value=0)

# print result
print(status_service)

# draw a chart to display results
status_service.plot(kind="bar")

plt.title ("Delivery Status by Service Type") # title for the chart

plt.ylabel ("Number of Deliveries") # print the y-axis

plt.xlabel ("Service Type") # print the x-axis
```
